```python
from sklearn.preprocessing import LabelEncoder, MinMaxScaler
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd

# Load dataset
df = pd.read_csv("myntradataset.csv")

# Handle missing values
df['brand'] = df['brand'].fillna('Unknown')
df['colour'] = df['colour'].fillna('Unknown')

# Encode categorical features
brand_enc = LabelEncoder()
colour_enc = LabelEncoder()
df['brand_encoded'] = brand_enc.fit_transform(df['brand'])
df['colour_encoded'] = colour_enc.fit_transform(df['colour'])

# Select features
features = df[['avg_rating', 'price', 'brand_encoded', 'colour_encoded']].fillna(0)

# Normalize
scaler = MinMaxScaler()
features_scaled = scaler.fit_transform(features)

# Function: Collaborative-Style Recommendation
def recommend_collaborative(product_index, top_n=5):
    sim_scores = cosine_similarity([features_scaled[product_index]], features_scaled).flatten()
    sim_indices = sim_scores.argsort()[-top_n-1:-1][::-1]
    recs = df.iloc[sim_indices][['name', 'brand', 'colour', 'price', 'avg_rating']].reset_index(drop=True)
    # Format output nicely
    output = ""
    for i, row in recs.iterrows():
        output += f"{i+1}. {row['name']}\n"
        output += f" Brand: {row['brand']}\n"
        output += f" Colour: {row['colour']}\n"
        output += f" Price: {row['price']}\n"
        output += f" Avg. Rating: {row['avg_rating']}\n\n"
    return output

# Example usage
print(recommend_collaborative(0, top_n=5))
```
