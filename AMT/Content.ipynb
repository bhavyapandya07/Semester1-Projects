```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Load dataset
df = pd.read_csv("myntradataset.csv")

# Fill missing values
df['description'] = df['description'].fillna('')
df['brand'] = df['brand'].fillna('Unknown')
df['colour'] = df['colour'].fillna('Unknown')

# Combine features for TF-IDF
df['combined_features'] = (df['description'] + " " + df['brand'] + " " + df['colour'])

# TF-IDF Vectorization
tfidf = TfidfVectorizer(stop_words='english', max_features=5000)
tfidf_matrix = tfidf.fit_transform(df['combined_features'])

# Function: Content-Based Recommendation
def recommend_content_based(product_index, top_n=5):
    sim_scores = cosine_similarity(tfidf_matrix[product_index], tfidf_matrix).flatten()
    sim_indices = sim_scores.argsort()[-top_n-1:-1][::-1]  # top N excluding itself
    recs = df.iloc[sim_indices][['name', 'brand', 'colour', 'price']].reset_index(drop=True)
    
    # Format output nicely
    output = ""
    for i, row in recs.iterrows():
        output += f"{i+1}. {row['name']}\n"
        output += f" Brand: {row['brand']}\n"
        output += f"Colour: {row['colour']}\n"
        output += f" Price: {row['price']}\n\n"
    return output

# Example usage
print("Content-Based Recommendations \n")
print(recommend_content_based(0, top_n=5))
```
